## 1. Execução manual do Selection Sort

Lista inicial:
[12, 25, 15, 3, 11, 24, 1, 9, 13, 16]

Resultados por iteração:

1ª: [1, 25, 15, 3, 11, 24, 12, 9, 13, 16]  
2ª: [1, 3, 15, 25, 11, 24, 12, 9, 13, 16]  
3ª: [1, 3, 9, 25, 11, 24, 12, 15, 13, 16]  
4ª: [1, 3, 9, 11, 25, 24, 12, 15, 13, 16]  
5ª: [1, 3, 9, 11, 12, 24, 25, 15, 13, 16]  
6ª: [1, 3, 9, 11, 12, 13, 25, 15, 24, 16]  
7ª: [1, 3, 9, 11, 12, 13, 15, 25, 24, 16]  
8ª: [1, 3, 9, 11, 12, 13, 15, 16, 24, 25]


```python
def selection_sort_2(lista):
    n = len(lista)

    for i in range(n):
        min_idx = i

        for j in range(i + 1, n):
            if lista[j] < lista[min_idx]:
                min_idx = j

        if min_idx != i:
            print(f"Troca: {lista[i]} por {lista[min_idx]}")
            lista[i], lista[min_idx] = lista[min_idx], lista[i]

        print(f"Resultado da {i+1}ª iteração - {lista}")

    return lista


lista = [12, 25, 15, 3, 11, 24, 1, 9, 13, 16]
selection_sort_2(lista)
```

## 3. Análise da lista ordenada

Não houve alteração na quantidade de comparações realizadas.  
O Selection Sort sempre percorre toda a lista para encontrar o menor elemento, mesmo quando ela já está ordenada.  
Sua complexidade permanece O(n²).


```python
selection_sort_2([1, 3, 9, 11, 12, 13, 15, 16, 24, 25])
```

```python
def verifica_ordenacao(lista):
    for i in range(len(lista) - 1):
        if lista[i] > lista[i + 1]:
            print("Lista não ordenada")
            return
    print("Lista ordenada")


verifica_ordenacao([1, 2, 3])
verifica_ordenacao([3, 2, 1])
```

```python
import matplotlib.pyplot as plt
import random
import time

def selection_sort(lista):
    n = len(lista)
    for i in range(n):
        min_idx = i
        for j in range(i + 1, n):
            if lista[j] < lista[min_idx]:
                min_idx = j
        lista[i], lista[min_idx] = lista[min_idx], lista[i]

tamanhos = []
tempos = []

for n in range(100, 1000, 100):
    lista = [random.randint(0, 1000) for _ in range(n)]

    inicio = time.time()
    selection_sort(lista)
    fim = time.time()

    tamanhos.append(n)
    tempos.append(fim - inicio)

plt.plot(
    tamanhos,
    tempos,
    marker='o',
    markersize=8,
    color='blue',
    markerfacecolor='red',
    label='Selection Sort'
)

plt.xlabel('Tamanho da lista')
plt.ylabel('Tempo de execução')
plt.title('Complexidade do Selection Sort')

plt.legend(loc='best')
plt.show()
```
