## Selection Sort - Ordenação por seleção

**Selection Sort** é um algoritmo de ordenação simples que funciona da seguinte forma:

1. Primeiro, encontramos o menor elemento e o trocamos com o primeiro elemento. Assim, colocamos o menor elemento em sua posição correta.

2. Então encontramos o menor entre os elementos restantes (ou o segundo menor) e o trocamos com o segundo elemento.

3. Os passos 1 e 2 são repetidos para o restante da lista.

Esse processo continua até que a lista esteja ordenada.

**Análise Assintótica**

O Selection Sort tem uma **complexidade de tempo de execução O(n<sup>2</sup>)** no pior caso, médio caso e melhor caso, o que o torna ineficiente para grandes conjuntos de dados.

Algoritmo estável: não

Requer memória auxiliar: não

```python
#Selection sort
def selection_sort(lista):
    n = len(lista)
    for i in range(n):
        min_index = i
        for j in range(i + 1, n):
            if lista[j] < lista[min_index]:
                min_index = j
        lista[i], lista[min_index] = lista[min_index], lista[i]
    return lista
```

```python
#teste do algoritmo
lista = [3, 1, 2, 7, 6]
print("Lista inicial:", lista)
lista_ordenada = selection_sort(lista)
print("Lista ordenada:", lista_ordenada)
```

## Insertion Sort - Ordenação por inserção

**Insertion Sort** é também um algoritmo de ordenação simples que funciona da seguinte forma:

1. Percorre a lista e considera cada elemento como uma "carta" a ser inserida em sua posição correta.
2. Compara cada elemento com os elementos anteriores e o insere na posição correta.
3. Repete o processo até que a lista esteja ordenada.

**Análise Assintótica**

O Insertion Sort tem uma **complexidade de tempo de O(n<sup>2</sup>)** no pior caso e médio caso, mas possui complexidade **O(n)** **no melhor caso** (com os dados de entrada já ordenados).

Algoritmo estável: sim

Requer memória auxiliar: não

```python
#Insertion sort
def insertion_sort(lista):
  n = len(lista)
  for i in range(1, n):
    chave = lista[i]
    j = i - 1
    while j >= 0 and lista[j] > chave:
      lista[j + 1] = lista[j]
      j -= 1
    lista[j + 1] = chave
  return lista
```

```python
#teste do algoritmo
lista = [3, 1, 2, 7, 6]
print("Lista inicial:", lista)
lista_ordenada = insertion_sort(lista)
print("Lista ordenada:", lista_ordenada)
```

## Bubble Sort - Ordenação bolha

**Bubble Sort** é um algoritmo de ordenação que funciona da seguinte forma:

1. Percorre a lista e compara cada par de elementos adjacentes.
2. Se os elementos estiverem na ordem errada, troca-os.
3. Repete o processo até que a lista esteja ordenada.

**Análise Assintótica**

O Bubble Sort tem uma **complexidade de tempo de O(n<sup>2</sup>)** no pior caso e no caso  médio. No melhor caso, quando a lista já está ordenada ou quase ordenada, a **complexidade pode ser O(n)**.

Algoritmo estável: sim

Requer memória auxiliar: não

```python
def bubble_sort(arr):
    a = arr.copy()
    n = len(a)
    for i in range(n):
        for j in range(0, n-i-1):
            if a[j] > a[j+1]:
                a[j], a[j+1] = a[j+1], a[j]
    return a
```

```python
#teste do algoritmo
lista = [3, 1, 2, 7, 6]
print("Lista inicial:", lista)
ordenada = bubble_sort(lista)
print("Lista ordenada (Bubble Sort):", ordenada)
```

## Merge Sort - Ordenação por junção

**Merge Sort** é um algoritmo de ordenação que utiliza a técnica de divisão e conquista (dividir para conquistar):

1. Divide a lista em duas metades.
2. Ordena cada metade recursivamente.
3. Combina as duas metades ordenadas em uma lista única ordenada.

**Análise Assintótica**

O Merge Sort tem uma **complexidade de tempo de O(n log n)**, o que o torna mais eficiente para grandes conjuntos de dados.

Algoritmo estável: sim

Requer memória auxiliar: sim

```python
def merge_sort(arr):
    if len(arr) <= 1:
        return arr
    mid = len(arr)//2
    left = merge_sort(arr[:mid])
    right = merge_sort(arr[mid:])
    return merge(left, right)

def merge(left, right):
    result = []
    i = j = 0
    while i < len(left) and j < len(right):
        if left[i] < right[j]:
            result.append(left[i])
            i += 1
        else:
            result.append(right[j])
            j += 1
    result.extend(left[i:])
    result.extend(right[j:])
    return result
```

```python
#teste do algoritmo
lista = [3, 1, 2, 7, 6]
print("Lista inicial:", lista)
ordenada = merge_sort(lista)
print("Lista ordenada (Merge Sort):", ordenada)
```

## Quick Sort - Ordenação rápida

**Quick Sort** é outro algoritmo de ordenação que utiliza a técnica de divisão e conquista:

1. Escolhe um elemento pivô. Pode ser o primeiro elemento, o último ou o do meio.
2. Divide a lista em duas partes: elementos menores que o pivô e elementos maiores que o pivô.
3. Ordena cada parte recursivamente.

**Análise Assintótica**

O Quick Sort tem uma **complexidade de O(n<sup>2</sup>) no pior caso**, porém apresenta **complexidade de tempo de O(n log n) no caso médio e melhor caso**, dependendo da escolha do pivô. O pior caso ocorre quando o pivô divide o array em uma partição "ruim", como uma metade com tamanho 0 e outra com tamanho n-1.

Algoritmo estável: não

Requer memória auxiliar: sim, mas requer somente memória adicional para a pilha de chamadas em implementações recursivas.

```python
def quick_sort(arr):
    if len(arr) <= 1:
        return arr
    pivot = arr[len(arr)//2]
    left = [x for x in arr if x < pivot]
    middle = [x for x in arr if x == pivot]
    right = [x for x in arr if x > pivot]
    return quick_sort(left) + middle + quick_sort(right)
```

```python
#teste do algoritmo
lista = [3, 1, 2, 7, 6]
print("Lista inicial:", lista)
ordenada = quick_sort(lista)
print("Lista ordenada (Quick Sort):", ordenada)
```

### Visualização da complexidade dos algoritmos de ordenação

```python
#Lista com os algoritmos utilizados
algoritmos = {
    "Selection Sort": selection_sort,
    "Insertion Sort": insertion_sort,
    "Bubble Sort": bubble_sort,
    "Merge Sort": merge_sort,
    "Quick Sort": quick_sort
}
```

```python
import matplotlib.pyplot as plt
import random
import time

#Função que gera uma lista aleatória de tamanho n
def gerar_lista(n):
    return [random.randint(0, 1000) for _ in range(n)]

#Função para medir o tempo de execução de um algoritmo
def medir_tempo(algoritmo, lista):
    inicio = time.time()
    algoritmo(lista)
    fim = time.time()
    return fim - inicio

#Tamanhos de lista
tamanhos = [100,200,300,400,500,600,700,800,900,1000]

#Código para medir o tempo de execução
tempos = {nome: [] for nome in algoritmos}

for nome, algoritmo in algoritmos.items():
    for tamanho in tamanhos:
        lista = gerar_lista(tamanho)
        tempo = medir_tempo(algoritmo, lista)
        tempos[nome].append(tempo)

#Gráfico
plt.figure(figsize=(10,6))
for nome in algoritmos:
    plt.plot(tamanhos, tempos[nome], marker='o', label=nome)

plt.xlabel("Tamanho da lista (n)")
plt.ylabel("Tempo de execução (segundos)")
plt.title("Comparação de complexidade dos algoritmos de ordenação")
plt.legend()
plt.grid(True)
plt.show()
```
